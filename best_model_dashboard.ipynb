```python
import pandas as pd
import matplotlib.pyplot as plt

df_1 = pd.read_csv("./best_model_experiment/experiments_v1/summary.csv").drop("run_id", axis=1)
df_2 = pd.read_csv("./best_model_experiment/experiments_v2/summary.csv").drop("run_id", axis=1)

for col in ["pr_auc", "recall_anomaly", "precision_anomaly", "f1_anomaly", "accuracy"]:
    if col in df_1.columns:
        df_1[col] = pd.to_numeric(df_1[col], errors="coerce")
        df_2[col] = pd.to_numeric(df_2[col], errors="coerce")

df_1
```

```python
agg_1 = (
    df_1.groupby(["sweep_key", "sweep_value"], as_index=False)
        .agg(
          pr_auc_mean=("pr_auc", "mean"),
          pr_auc_std=("pr_auc", "std"),
          recall_mean=("recall_anomaly", "mean"),
          recall_std=("recall_anomaly", "std"),
          f1_mean=("f1_anomaly", "mean"),
          f1_std=("f1_anomaly", "std"),
      )
)
agg_2 = (
    df_2.groupby(["sweep_key", "sweep_value"], as_index=False)
        .agg(
          pr_auc_mean=("pr_auc", "mean"),
          pr_auc_std=("pr_auc", "std"),
          recall_mean=("recall_anomaly", "mean"),
          recall_std=("recall_anomaly", "std"),
          f1_mean=("f1_anomaly", "mean"),
          f1_std=("f1_anomaly", "std"),
      )
)

agg_1 = agg_1.sort_values(["sweep_key", "sweep_value"])
agg_2 = agg_2.sort_values(["sweep_key", "sweep_value"])
agg_1
```

```python
agg_2
```

```python
group_cols = [
    "sweep_key",
    "sweep_value",
    "feature_map",
    "ansatz",
    "ansatz_depth",
    "optimizer",
    "n_samples",
    "anomaly_ratio",
    "label_noise"
]

metric_cols = [
    "pr_auc",
    "recall_anomaly",
    "precision_anomaly",
    "f1_anomaly",
    "accuracy",
    "tn",
    "fp",
    "fn",
    "tp"
]

df_avg = (
    df_2.groupby(group_cols)[metric_cols]
    .mean()
    .reset_index()
)

df_avg["seed"] = "avg"

df_2 = pd.concat([df_2, df_avg], ignore_index=True)
```
